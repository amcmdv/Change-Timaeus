**Project 1: The Gravity of Fate vs. Free Will**.

This script modifies your JSON structure to integrate the Greek concepts of divine bribery, the corrupting nature of desire, and the origin of fate. It introduces "Paris" and "Orestes" as new personas to contrast with your African kings, mapping how their actions align with either *Ma'at* (duty/order) or *Isfet* (hubris/chaos/trauma).


```
# 1. Install the required network physics library
!pip install pyvis -q

import json
from pyvis.network import Network
import IPython

# 2. Embed the integrated Ma'at + Greek Mythology data
data_json = """
{
  "name": "mꜣꜥt", "gloss": "Ma'at (Cosmic Order)", "category": "Spirituality", "epochs": ["Egypt", "Nubia", "Aksum", "Mali", "Greece"],
  "description": "The fundamental principle of existence: truth, balance, order, law, and justice.",
  "children": [
    {
      "name": "nsw", "gloss": "The King / Hero", "category": "Law/Duty", "epochs": ["Egypt", "Aksum", "Mali", "Greece"],
      "description": "The central pivot who maintains order, or destroys it through hubris.",
      "children": [
        {
          "name": "hp", "gloss": "Law/Custom", "category": "Law", "epochs": ["Egypt", "Mali", "Greece"],
          "description": "The codification of social rules and family duty.",
          "children": [
            {"name": "Kouroukan Fouga", "gloss": "Manden Charter", "category": "Social", "epochs": ["Mali"], "description": "Sundiata's constitution declaring social peace."},
            {"name": "Moral Guilt", "gloss": "Justice vs Vengeance", "category": "Psychology", "epochs": ["Greece"], "description": "Orestes' paralyzing decision to avenge his father, hunted by the Furies."}
          ]
        }
      ]
    },
    {
      "name": "isfet", "gloss": "Isfet (Chaos / Tragedy)", "category": "Spirituality", "epochs": ["Egypt", "Mali", "Aksum", "Greece"],
      "description": "The opposite of Ma'at: violence, injustice, lies, and unchecked emotion.",
      "children": [
        {"name": "Sumanguru", "gloss": "Sorcerer King", "category": "Social", "epochs": ["Mali"], "description": "The antagonist to Sundiata; represented chaos and tyranny."},
        {
          "name": "Hubris", "gloss": "Destructive Desire", "category": "Psychology", "epochs": ["Greece"],
          "description": "The corrupting nature of desire and the origin of fate.",
          "children": [
            {"name": "Judgment of Paris", "gloss": "Decision Paralysis", "category": "Fate", "epochs": ["Greece"], "description": "Divine bribery and the psychological burden of a choice affecting the world."},
            {"name": "Intergenerational Trauma", "gloss": "Blood Guilt", "category": "Fate", "epochs": ["Greece"], "description": "The concept that crime necessitates retribution, passing suffering down the lineage."}
          ]
        }
      ]
    }
  ]
}
"""

# Added Greek Personas mapping to the intense psychological conflicts
personas_json = """
{
  "ezana": {"color": "#00bcd4", "name": "Ezana of Aksum", "nodes": ["nsw", "mꜣꜥt"]},
  "sundiata": {"color": "#ffc107", "name": "Sundiata Keita", "nodes": ["Kouroukan Fouga", "Sumanguru", "hp", "nsw", "mꜣꜥt"]},
  "paris": {"color": "#ff5722", "name": "Paris of Troy", "nodes": ["Judgment of Paris", "Hubris", "isfet"]},
  "orestes": {"color": "#8bc34a", "name": "Orestes", "nodes": ["Moral Guilt", "Intergenerational Trauma", "hp", "isfet"]}
}
"""

data = json.loads(data_json)
personas = json.loads(personas_json)

# 3. Initialize the Pyvis Network with a dark theme
net = Network(height="700px", width="100%", bgcolor="#121212", font_color="#e0e0e0", notebook=True, cdn_resources='remote')

# PHYSICS TWEAK: Increased spring length and lowered gravity to visualize the "repulsion" of chaotic Greek elements
net.force_atlas_2based(gravity=-85, central_gravity=0.005, spring_length=200)

sundiata_nodes = set(personas["sundiata"]["nodes"])
ezana_nodes = set(personas["ezana"]["nodes"])
paris_nodes = set(personas["paris"]["nodes"])
orestes_nodes = set(personas["orestes"]["nodes"])

# 4. Recursive function to parse JSON and map it to graph nodes/edges
def add_nodes_edges(node, parent_name=None):
    node_id = node['name']
    label = f"{node['name']}\n({node['gloss']})"
    title = f"Epochs: {', '.join(node.get('epochs', []))}\nCategory: {node.get('category', 'N/A')}\n\n{node.get('description', '')}"

    # Complex color logic for multiple intersections
    node_color = "#333333" # Default
    if node_id == "mꜣꜥt":
        node_color = "#ffffff" # Absolute order is white
    elif node_id == "isfet":
        node_color = "#e53935" # Absolute chaos is red
    elif node_id in paris_nodes:
        node_color = personas["paris"]["color"]
    elif node_id in orestes_nodes:
        node_color = personas["orestes"]["color"]
    elif node_id in sundiata_nodes:
        node_color = personas["sundiata"]["color"]
    elif node_id in ezana_nodes:
        node_color = personas["ezana"]["color"]

    node_size = 25 if not parent_name else 15

    # Added mass to Greek tragedy nodes to make them violently disrupt the graph's balance
    node_mass = 5 if node_id in ["Judgment of Paris", "Intergenerational Trauma"] else 1

    net.add_node(node_id, label=label, title=title, color=node_color, size=node_size, shape="dot", mass=node_mass)

    if parent_name:
        # Greek trauma nodes get red springs to indicate tension
        edge_color = "#ff5252" if node_id in paris_nodes or node_id in orestes_nodes else "#444444"
        net.add_edge(parent_name, node_id, color=edge_color)

    if 'children' in node:
        for child in node['children']:
            add_nodes_edges(child, node_id)

add_nodes_edges(data)

net.show_buttons(filter_=['physics'])
net.show("maat_myth_network.html")
IPython.display.HTML("maat_myth_network.html")
```

### Key Changes to Notice:

* **Physics Tweaks:** `central_gravity` lowered to `0.005` and increased the `spring_length` to `200`. This allows the "Isfet/Chaos" side of the graph to physically push further away from the "Ma'at/Order" core, visually simulating the fracturing of cosmic order by human tragedy.


* **Mass Injection:** The `node_mass = 5` logic gives specific Greek tragedy nodes (like *Judgment of Paris*) heavier physical weight. In Pyvis, heavier nodes pull their connected nodes more aggressively, effectively visualising the psychological burden of Paris's choice.


* **Edge Colors:** Edges connecting to the Greek mythological trauma are tinted red (`#ff5252`) to represent blood guilt and the violent spiral of retribution.
