**Project 2: Visualising complex intersections—Structural order of Ma'at with Family trees of Greek tragedy**.

Managing data density is a fantastic usability requirement. When visualising complex intersections—especially when mixing the structural order of Ma'at with the chaotic family trees of Greek tragedy—the graph can quickly become overwhelming.

To solve this, I have updated the codebase with two specific usability upgrades:

1. **Programmatic Toggles:** A simple `ACTIVE_MODULES` dictionary at the very top of the script allows the user to turn specific mythological cycles (like the Theban cycle or the Cycle of Atreus) `True` or `False` before rendering.
2. **Pyvis Native Filtering:** I enabled Pyvis's `select_menu=True` and `filter_menu=True`. This adds native HTML dropdowns to the final output, allowing the user to isolate specific nodes or categories (e.g., isolating only "Psychology" or "Law" nodes) in real-time.

The enriched codebase now fully integrates the stories focused on Achilles' rage, Clytemnestra's vengeance, and Oedipus's self-discovery:

```
# 1. Install the required network physics library
!pip install pyvis -q

import json
from pyvis.network import Network
import IPython

# =====================================================================
# USABILITY TOGGLES: Turn specific mythological cycles True/False
# to control how dense the visualization is before generating.
# =====================================================================
ACTIVE_MODULES = {
    "african_history": True, # Sundiata, Ezana, Ma'at baseline
    "greek_troy": True,      # Paris, Achilles (The Trojan War Cycle)
    "greek_atreus": True,    # Agamemnon, Orestes (The Cycle of Atreus)
    "greek_thebes": False    # Oedipus, Antigone (The Theban Cycle) - Toggled off by default to reduce clutter
}

# 2. Embed the highly enriched Ma'at + Greek Mythology data
data_json = """
{
  "name": "mꜣꜥt", "gloss": "Ma'at (Cosmic Order)", "category": "Spirituality", "epochs": ["Egypt", "Nubia", "Aksum", "Mali", "Greece"],
  "description": "The fundamental principle of existence: truth, balance, order, law, and justice.",
  "children": [
    {
      "name": "nsw", "gloss": "The King / Hero", "category": "Law/Duty", "epochs": ["Egypt", "Aksum", "Mali", "Greece"],
      "description": "The central pivot who maintains order, or destroys it through hubris.",
      "children": [
        {
          "name": "hp", "gloss": "Law/Custom", "category": "Law", "epochs": ["Egypt", "Mali", "Greece"],
          "description": "The codification of social rules and family duty.",
          "children": [
            {"name": "Kouroukan Fouga", "gloss": "Manden Charter", "category": "Social", "epochs": ["Mali"], "description": "Sundiata's constitution declaring social peace.", "module": "african_history"},
            {"name": "Moral Guilt", "gloss": "Justice vs Vengeance", "category": "Psychology", "epochs": ["Greece"], "description": "Orestes' paralyzing decision to avenge his father, hunted by the Furies.", "module": "greek_atreus"},
            {"name": "Ethical Absolutism", "gloss": "Conscience vs Law", "category": "Psychology", "epochs": ["Greece"], "description": "Antigone's uncompromising adherence to moral principle despite fatal consequences.", "module": "greek_thebes"}
          ]
        }
      ]
    },
    {
      "name": "isfet", "gloss": "Isfet (Chaos / Tragedy)", "category": "Spirituality", "epochs": ["Egypt", "Mali", "Aksum", "Greece"],
      "description": "The opposite of Ma'at: violence, injustice, lies, and unchecked emotion.",
      "children": [
        {"name": "Sumanguru", "gloss": "Sorcerer King", "category": "Social", "epochs": ["Mali"], "description": "The antagonist to Sundiata; represented chaos and tyranny.", "module": "african_history"},
        {
          "name": "Hubris", "gloss": "Destructive Desire", "category": "Psychology", "epochs": ["Greece"],
          "description": "The corrupting nature of desire and the origin of fate.",
          "children": [
            {"name": "Judgment of Paris", "gloss": "Decision Paralysis", "category": "Fate", "epochs": ["Greece"], "description": "Divine bribery and the psychological burden of a choice affecting the world.", "module": "greek_troy"},
            {"name": "Rage of Achilles", "gloss": "Unchecked Emotion", "category": "Fate", "epochs": ["Greece"], "description": "The psychological toll of hubris and the destructive power of unchecked emotion.", "module": "greek_troy"},
            {"name": "Intergenerational Trauma", "gloss": "Blood Guilt", "category": "Fate", "epochs": ["Greece"], "description": "The concept that crime necessitates retribution, passing suffering down the lineage.", "module": "greek_atreus"},
            {"name": "Trauma of Self-Discovery", "gloss": "Blindness to Truth", "category": "Psychology", "epochs": ["Greece"], "description": "The psychological refusal to see the obvious truth, leading to profound trauma.", "module": "greek_thebes"}
          ]
        }
      ]
    }
  ]
}
"""

personas_json = """
{
  "ezana": {"color": "#00bcd4", "name": "Ezana of Aksum", "nodes": ["nsw", "mꜣꜥt"], "module": "african_history"},
  "sundiata": {"color": "#ffc107", "name": "Sundiata Keita", "nodes": ["Kouroukan Fouga", "Sumanguru", "hp", "nsw", "mꜣꜥt"], "module": "african_history"},
  "paris": {"color": "#ff5722", "name": "Paris of Troy", "nodes": ["Judgment of Paris", "Hubris", "isfet"], "module": "greek_troy"},
  "achilles": {"color": "#d84315", "name": "Achilles", "nodes": ["Rage of Achilles", "Hubris", "nsw", "isfet"], "module": "greek_troy"},
  "agamemnon": {"color": "#673ab7", "name": "Agamemnon & Clytemnestra", "nodes": ["Intergenerational Trauma", "hp", "isfet"], "module": "greek_atreus"},
  "orestes": {"color": "#8bc34a", "name": "Orestes", "nodes": ["Moral Guilt", "Intergenerational Trauma", "hp", "isfet"], "module": "greek_atreus"},
  "oedipus": {"color": "#795548", "name": "Oedipus", "nodes": ["Trauma of Self-Discovery", "hp", "isfet"], "module": "greek_thebes"},
  "antigone": {"color": "#c2185b", "name": "Antigone", "nodes": ["Ethical Absolutism", "hp", "mꜣꜥt"], "module": "greek_thebes"}
}
"""

data = json.loads(data_json)
personas = json.loads(personas_json)

# Compile a set of active node names based on our toggles
active_nodes = {"mꜣꜥt", "nsw", "hp", "isfet", "Hubris"} # Base structural nodes
for persona_key, persona_data in personas.items():
    if ACTIVE_MODULES.get(persona_data.get("module", ""), True):
        active_nodes.update(persona_data["nodes"])

# 3. Initialize Pyvis with USABILITY UI TURNED ON
net = Network(
    height="800px",
    width="100%",
    bgcolor="#121212",
    font_color="#e0e0e0",
    notebook=True,
    cdn_resources='remote',
    select_menu=True, # Usability: Adds a dropdown to select specific nodes
    filter_menu=True  # Usability: Adds a dropdown to filter by category/group
)

net.force_atlas_2based(gravity=-90, central_gravity=0.005, spring_length=220)

# 4. Recursive mapping function with density checks
def add_nodes_edges(node, parent_name=None):
    node_id = node['name']

    # DENSITY CHECK: Skip this node and its children if its module is toggled off
    node_module = node.get('module')
    if node_module and not ACTIVE_MODULES.get(node_module, True):
        return

    label = f"{node['name']}\n({node['gloss']})"
    category = node.get('category', 'Concept')
    title = f"Epochs: {', '.join(node.get('epochs', []))}\nCategory: {category}\n\n{node.get('description', '')}"

    # Determine colors based on active personas
    node_color = "#333333"
    if node_id == "mꜣꜥt": node_color = "#ffffff"
    elif node_id == "isfet": node_color = "#e53935"
    else:
        for p_key, p_data in personas.items():
            if ACTIVE_MODULES.get(p_data.get("module"), True) and node_id in p_data["nodes"]:
                node_color = p_data["color"]
                break

    node_size = 30 if not parent_name else 18
    node_mass = 5 if category in ["Fate", "Psychology"] else 1 # Greek nodes pull harder

    # Add node to graph, grouping by category for the filter menu
    net.add_node(node_id, label=label, title=title, color=node_color, size=node_size, mass=node_mass, group=category)

    if parent_name:
        edge_color = "#ff5252" if category in ["Fate", "Psychology"] else "#444444"
        net.add_edge(parent_name, node_id, color=edge_color)

    if 'children' in node:
        for child in node['children']:
            add_nodes_edges(child, node_id)

add_nodes_edges(data)

# 5. Generate inline
net.show_buttons(filter_=['physics'])
net.show("maat_myth_network_interactive.html")
IPython.display.HTML("maat_myth_network_interactive.html")
```

The new dropdown menus at the top of the output cell. They allow you to select a specific node to highlight it, or filter the graph by `group` (which I have mapped to the data's `category`, like "Psychology" or "Social").

This gives you both macro-control (Python `True/False` variables) and micro-control (UI dropdowns) over the visualisation's density.
